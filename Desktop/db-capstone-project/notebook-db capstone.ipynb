```python
# Task 3: Query to retrieve customer details for orders over $60
customer_query = """
SELECT Customer.FullName, Customer.Email, Customer.ContactNumber, Orders.TotalCost
FROM Customer
INNER JOIN Orders ON Customer.CustomerID = Orders.Customer_CustomerID
WHERE Orders.TotalCost > 60;
"""

# Execute the query
cursor.execute(customer_query)

# Fetch the results
customer_results = cursor.fetchall()

# Print the results (full name, contact details, and total cost)
print("Customers with orders over $60:")
for result in customer_results:
    full_name,email, phone, total_cost = result
    print(f"Name: {full_name} , Email: {email}, Phone: {phone}, TotalCost: ${total_cost}")
```
